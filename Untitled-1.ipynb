```python
# 🌐 Symbolic AI Setup
from sympy import *
init_printing(use_unicode=True)

# 📌 Declare symbol(s)
x, y, z = symbols('x y z')

# 🧪 Sample Test Equation (like from Mobius)
expr = log(x**2)

# 🔍 Run transformation
simplified_expr = simplify(expr)

# 📤 Output results
display(expr)
display(Eq(expr, simplified_expr, evaluate=False))
```
